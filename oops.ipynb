## CLASS
A class is a blueprint for creating objects. It defines a set of attributes and methods that the objects created from the class will have. In Python, a class is defined using the `class` keyword. For example:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."
```


## class : data/attributes and function/methods
Attributes are the variables that hold data specific to an object, while methods are functions defined within a class that operate on the object's data. For example, in the list `L = [2, 3, 3, 4, 12]`, `L` is an object of the `list` class. It has attributes like its length (`len(L)`) and methods like `append()`, `pop()`, and `sort()`.
